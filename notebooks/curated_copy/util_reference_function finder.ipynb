```python
import pkgutil, importlib

import torchvision

pkg, name = torchvision, "Image"
out = []
for m in pkgutil.walk_packages(pkg.__path__, pkg.__name__ + "."):
    try: mod = importlib.import_module(m.name)
    except Exception: continue
    if any(a.lower() == name.lower() for a in dir(mod)): #hasattr(mod, name):
        out.extend([m.name +"." + a for a in dir(mod) if a.lower().find(name.lower()) > -1 ])
        # if len(out)>=n: break
out[:3]
```
